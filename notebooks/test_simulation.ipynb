```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np

sys.path.append(str(Path.cwd().parent))

from memristor_simulation import MemristorSimulator
```

```python
g_start_us = 220.0
voltage = 1.2
duration_ns = 5000.0
sample_every_ns = 10.0
n_repeats = 128

simulator = MemristorSimulator()
_ = simulator.backend
```

```python
simulator.reset(g_start_us)
trace = simulator.simulate_constant_voltage(
    voltage=voltage,
    duration_ns=duration_ns,
    sample_every_ns=sample_every_ns,
)
trace.conductances_us[-1]
```

```python
plt.figure(figsize=(8, 4))
plt.plot(trace.times_ns, trace.conductances_us)
plt.xlabel('time_ns')
plt.ylabel('conductance_us')
plt.tight_layout()
```

```python
def run_once():
    simulator.reset(g_start_us)
    return simulator.simulate_constant_voltage(
        voltage=voltage,
        duration_ns=duration_ns,
        sample_every_ns=sample_every_ns,
    )

traces = [run_once() for _ in range(n_repeats)]
final_conductances = np.array([item.conductances_us[-1] for item in traces])
```

```python
plt.figure(figsize=(8, 4))
for item in traces[:16]:
    plt.plot(item.times_ns, item.conductances_us, alpha=0.5)
plt.xlabel('time_ns')
plt.ylabel('conductance_us')
plt.tight_layout()
```

```python
plt.figure(figsize=(8, 4))
plt.hist(final_conductances, bins=24)
plt.xlabel('final_conductance_us')
plt.ylabel('count')
plt.tight_layout()
```
